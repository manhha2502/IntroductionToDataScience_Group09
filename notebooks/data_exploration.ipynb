## **Bước 1: Xử lý kiểu dữ liệu**

- Đọc file CSV gốc
- Chuyển đổi các cột phần trăm (%) sang số thập phân
- Chuẩn hóa kiểu dữ liệu cho các cột số (float64)
- Lưu file đã xử lý vào `../data/processed/data_typed.csv`


```python
input_path = "../data/raw/data.csv"
output_path = "../data/processed/data_typed.csv"

df = process_data_types(input_path, output_path)

print(f"\nĐã xử lý kiểu dữ liệu xong!")
print(f"DataFrame {df.shape}")
```

## **Bước 2: Thống kê mô tả**


- Tính toán các thống kê mô tả cho các cột số quan trọng:
    - **Mean**
    - **Median**
    - **Std**
    - **Min**
    - **Max**
- Định dạng hiển thị 
    - **Tiền tệ (VND)**: hiển thị dấu phẩy ngăn cách hàng nghìn
    - **Tỉ lệ**: hiển thị 2 chữ số thập phân
    - **Số đếm**: hiển thị số nguyên


```python
numerical_cols = [
    'price', 'original_price', 'quantity_sold',
    'review_count', 'store_review_count', 'total_follower', 
    'rating_average', 'discount_rate', 'image_count', 'video_count'
]

stats_df = calculate_descriptive_stats(df, numerical_cols)

print("\nBẢNG THỐNG KÊ MÔ TẢ:")
print("="*60)
display(stats_df)
```

## **Bước 3: Trực quan hóa dữ liệu**

Tạo các biểu đồ:
- **Histogram**: Phân phối của 10 thuộc tính số quan trọng (có log transformation cho một số cột)
- **Heatmap**: Ma trận tương quan giữa các biến số
- **Bar Chart**: Thành phần sản phẩm theo danh mục cấp cao

Các file hình ảnh được lưu trong `../data/images/`


```python
output_dir = "../data/images"

print("Đang tạo histogram phân phối...")
plot_all_numerical_distributions(df, numerical_cols, output_dir)

print("\nĐang tạo heatmap tương quan...")
plot_correlation_heatmap(df, numerical_cols, output_dir)

print("\nĐang tạo bar chart thành phần danh mục...")
plot_category_composition(df, output_dir)

print("\nĐã tạo xong tất cả các biểu đồ")
```
